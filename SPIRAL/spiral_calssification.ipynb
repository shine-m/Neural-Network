```python

class dense_layer:
    def __init__(self, n_input, n_neuron):
        self.weights=np.random.randn(n_input, n_neuron)
        self.weights *= .01 #to keep the numbers small
        # .01 will be multiplied to all the entries of the weight matrix
        self.bias = np.zeros(n_neuron)

    def forward (self, x):
        self.output= np.dot(x, self.weights)+ self.bias
```

<h1>Activation_ReLU</h1>

```python
class activation_ReLU:
    def forward(self,x):
        self.output = np.maximum(0, x)
```

# Activation_softmax

```python
class Activation_softmax:
    def forward(self,x):
        self.output = x-np.max(x, axis=1, keepdims=1)
        self.output = np.exp(self.output)
        self.output = self.output/ np.sum(self.output, axis=1, keepdims=1)
```

# Layer 1 of neurons

```python
dense1 = dense_layer(2,3)
x,y= spiral_data(classes=3, samples= 100)
print(x[:3])
dense1.forward(x)

print("first layer output\n",dense1.output[:5])

act1 = activation_ReLU()
act1.forward(dense1.output)

final_output_from_layer_1 = act1.output




```

# layer 2 neurons

```python

dense2 = dense_layer(3,3)
dense2.forward(final_output_from_layer_1)
print("matrix multiplication \n",dense2.output[:3])

act2 = Activation_softmax()
act2.forward(dense2.output)
print("final output\n", act2.output[:4])
```
