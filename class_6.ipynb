creating dense layer class


```python
import numpy as np


class dense_layer:
    def __init__(self, n_inputs,n_neurons):
        self.weights = np.random.randn(n_inputs, n_neurons)*.01
        # to create n_inputs x n_neurons weight matrix
        self.bias = np.zeros((1,n_neurons))

    def forward (self ,inputs):
        self.output = np.dot(inputs, self.weights)+ self.bias



```

```python
import nnfs
from nnfs.datasets import spiral_data
nnfs.init()
x,y = spiral_data(samples=100, classes =3)

dense1 = dense_layer(2,3)
dense1.forward(x)

print(dense1.output[:10])
```

Activation functon (RELU)

```python
class Activation_Relu : 
    def forward(self, inputs):
        self.output= np.maximum(0, inputs)
        
```

```python
obj = Activation_Relu()
obj.forward(x)
print(obj.output[:5])
```

Softmax Activation Function

```python
class Activation_softmax:
    def forward(self, inputs):
        self.output = np.exp(inputs - np.max(inputs, axis =1, keepdims=1))
        self.probabilities = self.output/ (np.sum(self.output, axis =1, keepdims=1))
```

```python
obj = Activation_softmax    ()
obj.forward(dense1.output)
print(dense1.output[:5])
print(obj.probabilities[:5])
```
